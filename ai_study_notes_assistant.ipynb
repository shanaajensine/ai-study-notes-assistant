## Day 2 Notes: Text Vectorization and Similarity Search

Today I converted study note chunks into numerical form using TF-IDF.

### What I learned

- TF-IDF gives importance to meaningful words.

```python
study_notes = """
Machine learning is a field of artificial intelligence that allows computers to learn from data without being explicitly programmed.

Supervised learning is a type of machine learning where the model learns from labeled data. Labeled data means the input already has the correct output.

Regression is used when the target value is numerical. For example, predicting house prices, student marks, or temperature.

Classification is used when the target value is a category. For example, predicting whether an email is spam or not spam.

Unsupervised learning is used when the data does not have labels. The model tries to find patterns or groups in the data.

Clustering is an unsupervised learning technique that groups similar data points together.

A training set is used to teach the model. A test set is used to check how well the model performs on unseen data.

Overfitting happens when a model performs very well on training data but poorly on new data.

Underfitting happens when a model is too simple and performs poorly on both training and testing data.
"""

sentences = study_notes.split(".")

clean_sentences = []

for sentence in sentences:
    sentence = sentence.strip()
    if sentence:
        clean_sentences.append(sentence)

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()
note_vectors = vectorizer.fit_transform(clean_sentences)

def answer_question(question):
    question_vector = vectorizer.transform([question])
    similarity_scores = cosine_similarity(question_vector, note_vectors)
    best_match_index = similarity_scores.argmax()
    answer = clean_sentences[best_match_index]

    print("Question:", question)
    print("Answer:", answer)

print("Setup complete. You can continue Day 3.")
```

```python
def show_top_matches(question, top_n=3):
    question_vector = vectorizer.transform([question])
    similarity_scores = cosine_similarity(question_vector, note_vectors)[0]

    top_indices = similarity_scores.argsort()[::-1][:top_n]

    print("Question:", question)
    print()

    for rank, index in enumerate(top_indices, start=1):
        print("Rank:", rank)
        print("Score:", similarity_scores[index])
        print("Chunk:", clean_sentences[index])
        print("-----------------------------")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def process_notes(notes):
    chunks = notes.split("\n\n")

    clean_chunks = []

    for chunk in chunks:
        chunk = chunk.strip()
        if chunk:
            clean_chunks.append(chunk)

    vectorizer = TfidfVectorizer()
    note_vectors = vectorizer.fit_transform(clean_chunks)

    return clean_chunks, vectorizer, note_vectors
```

```python
sample_notes = """
Artificial Intelligence is a branch of computer science that focuses on creating machines that can perform tasks that normally require human intelligence.

Machine Learning is a part of artificial intelligence where computers learn patterns from data without being explicitly programmed.

Supervised learning is a type of machine learning where the model learns from labeled data.

Regression is used when the output value is numerical.

Classification is used when the output value is a category.

Overfitting happens when a model performs very well on training data but poorly on new data.

Neural networks are machine learning models inspired by the human brain. They are made of layers of connected nodes called neurons.
"""
```

```python
chunks, vectorizer, note_vectors = process_notes(sample_notes)

print("Notes processed successfully.")
print("Number of chunks:", len(chunks))
```

```python
def answer_question_from_notes(question, chunks, vectorizer, note_vectors, threshold=0.5):
    question_vector = vectorizer.transform([question])
    similarity_scores = cosine_similarity(question_vector, note_vectors)

    best_match_index = similarity_scores.argmax()
    best_score = similarity_scores[0][best_match_index]
    answer = chunks[best_match_index]

    print("Question:", question)
    print("Similarity Score:", best_score)

    if best_score < threshold:
        print("Answer: I could not find a good answer in the study notes.")
    else:
        print("Answer:", answer)
```

When the threshold was too low, the assistant gave an incorrect answer for a question outside the notes. Increasing the threshold made the assistant more careful and reduced weak matches.

```python
import gradio as gr

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def process_notes(notes):
    chunks = notes.split("\n\n")

    clean_chunks = []

    for chunk in chunks:
        chunk = chunk.strip()
        if chunk:
            clean_chunks.append(chunk)

    vectorizer = TfidfVectorizer()
    note_vectors = vectorizer.fit_transform(clean_chunks)

    return clean_chunks, vectorizer, note_vectors
```

```python
def get_answer(notes, question, threshold=0.5):
    if not notes.strip():
        return "Please enter study notes first.", 0.0

    if not question.strip():
        return "Please enter a question.", 0.0

    chunks, vectorizer, note_vectors = process_notes(notes)

    question_vector = vectorizer.transform([question])
    similarity_scores = cosine_similarity(question_vector, note_vectors)

    best_match_index = similarity_scores.argmax()
    best_score = similarity_scores[0][best_match_index]
    answer = chunks[best_match_index]

    if best_score < threshold:
        return "I could not find a good answer in the study notes.", round(best_score, 3)
    else:
        return answer, round(best_score, 3)
```

```python
default_notes = """
Artificial Intelligence is a branch of computer science that focuses on creating machines that can perform tasks that normally require human intelligence.

Machine Learning is a part of artificial intelligence where computers learn patterns from data without being explicitly programmed.

Supervised learning is a type of machine learning where the model learns from labeled data.

Regression is used when the output value is numerical.

Classification is used when the output value is a category.

Overfitting happens when a model performs very well on training data but poorly on new data.

Neural networks are machine learning models inspired by the human brain. They are made of layers of connected nodes called neurons.
"""
```

```python
app = gr.Interface(
    fn=get_answer,
    inputs=[
        gr.Textbox(
            label="Study Notes",
            lines=12,
            value=default_notes,
            placeholder="Paste your study notes here..."
        ),
        gr.Textbox(
            label="Question",
            placeholder="Ask a question from your notes..."
        )
    ],
    outputs=[
        gr.Textbox(label="Answer"),
        gr.Number(label="Similarity Score")
    ],
    title="AI Study Notes Assistant",
    description="Paste your study notes, ask a question, and get the most relevant answer using TF-IDF and cosine similarity."
)
```

```python
app.launch(share=True)
```

Students often have long study notes and may want quick answers from them. This project helps users ask questions from their own notes and receive the most relevant answer.

## How the Assistant Works

1. The user enters study notes.
2. The notes are split into smaller chunks.
3. The chunks are converted into numerical vectors using TF-IDF.
4. The user asks a question.
5. The question is also converted into a vector.
6. Cosine similarity compares the question with all note chunks.
7. The assistant returns the most relevant chunk as the answer.
8. If the similarity score is too low, the assistant says it could not find a good answer.


## Key Learnings

- Text data must be converted into numbers before a computer can compare it.
- TF-IDF gives importance to meaningful words in text.
- Cosine similarity helps find the most relevant note chunk for a question.
- Chunking affects answer quality.
- Paragraph chunks can give more complete answers than sentence chunks.
- A confidence threshold helps prevent weak or incorrect answers.
- Gradio can be used to create a simple app interface from Python code.


## Limitations

- This assistant does not truly understand language like a large language model.
- It uses keyword and similarity matching, so it may fail for complex questions.
- If the notes are too short, the similarity score may be unreliable.
- The assistant can only answer from the provided notes.
- It does not generate new explanations beyond the retrieved note chunk.
